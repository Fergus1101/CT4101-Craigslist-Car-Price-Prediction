```python
import os
import pandas as pd

# Define relative paths based on running the notebook inside the 'notebooks/' folder
# Use '../' to step up one directory to the project root
raw_data_path = "../data/raw/vehicles.csv"
output_data_path = "../data/processed/craigslist_30k_sample.csv"

# 1. Load the full raw dataset
print("Loading raw Kaggle dataset... (this may take a few seconds)")
df_raw = pd.read_csv(raw_data_path)
print(f"Full dataset loaded. Shape: {df_raw.shape}")

# 2. Draw a random sample of 30,000 records reproducibly
df_sample = df_raw.sample(n=30000, random_state=42)

# 3. Ensure the output directory exists and export the sample
os.makedirs("../data/processed", exist_ok=True)
df_sample.to_csv(output_data_path, index=False)

print(f"Sampling complete! Saved working sample to: {output_data_path}")
print(f"Sample shape: {df_sample.shape}")
```
