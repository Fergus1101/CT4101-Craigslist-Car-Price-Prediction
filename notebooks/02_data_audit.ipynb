```python
import pandas as pd
import numpy as np

# Load the working sample created in Step 2
df = pd.read_csv("../data/processed/craigslist_30k_sample.csv")

print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

```python
df.info()
```

```python
# Summary statistics for target variable
print("--- Price Statistics ---")
print(df['price'].describe())

# Count zero and extremely low/high price entries
zero_prices = (df['price'] == 0).sum()
under_500 = (df['price'] < 500).sum()
over_100k = (df['price'] > 100000).sum()

print(f"\nZero price entries: {zero_prices}")
print(f"Prices under $500: {under_500}")
print(f"Prices over $100,000: {over_100k}")
```

```python
# Check exact row duplicates
duplicate_rows = df.duplicated().sum()
print(f"Exact Duplicate Rows: {duplicate_rows}")

# Calculate percentage of missing values per column
missing_pct = (df.isna().mean() * 100).sort_values(ascending=False)
print("\n--- Missingness Percentage Per Column ---")
print(missing_pct[missing_pct > 0].round(2).astype(str) + '%')
```
